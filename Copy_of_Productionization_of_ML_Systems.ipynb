# **Travel Capstone Project Summary**

The Travel Capstone Project focuses on analyzing and integrating multiple aspects of the travel industry, combining data related to flights, hotels, and users to gain actionable insights and build potential predictive models. With the rapid growth of the tourism sector and increasing customer expectations for personalized experiences, businesses must leverage data to improve services, forecast demand, and enhance customer satisfaction. This project provides an end-to-end data science workflow, starting from data exploration to model development and deployment, using three core datasets: **flights**, **hotels**, and **users**.

The **flights dataset** contains detailed information about flight bookings, including attributes such as flight IDs, departure and arrival locations, travel dates, flight durations, prices, and booking statuses. This data enables the analysis of travel patterns, seasonal demand trends, popular routes, and fare fluctuations. By studying these patterns, businesses such as airlines and travel agencies can optimize their pricing strategies, improve operational efficiency, and predict future demand.

The **hotels dataset** provides information on hotel bookings, including hotel IDs, locations, room types, booking dates, check-in/check-out dates, total price, and ratings or reviews. This data allows for a deeper understanding of customer preferences for accommodations, identifying high-performing hotels, and evaluating booking behavior across different cities or regions. Combining this dataset with the flights data enables cross-analysis, such as identifying which flight destinations lead to the highest hotel booking demand or creating travel package recommendations.

The **users dataset** captures demographic and behavioral data about travelers, such as user IDs, names, age groups, location, travel preferences, and possibly loyalty program participation. This data is essential for building personalized experiences, segmenting customers into meaningful clusters, and targeting specific groups with tailored offers. For instance, young solo travelers might prefer budget airlines and hostels, while families may opt for premium flights and hotels with larger accommodations.

The primary goal of this project is to integrate these datasets and uncover relationships between user behavior, flight bookings, and hotel preferences. Exploratory Data Analysis (EDA) will be conducted to clean and visualize the data, identify missing values or anomalies, and reveal patterns. Key insights will include identifying peak travel seasons, predicting flight and hotel demand, and understanding customer booking cycles. These insights will be valuable for travel companies aiming to optimize inventory management and marketing campaigns.

From a machine learning perspective, the project has several potential directions. One key application could be **predictive modeling**, such as forecasting flight and hotel demand for upcoming months based on historical trends. Another possibility is **recommendation systems**, where personalized travel packages are suggested to users by analyzing their past behavior and comparing it with similar customer profiles. Classification models can also be developed to predict whether a booking is likely to be canceled or completed, helping businesses minimize revenue loss due to cancellations.

The tools and technologies used in this project include Python for data analysis, Pandas and NumPy for data manipulation, Matplotlib and Seaborn for visualization, and machine learning libraries such as Scikit-learn for model building. Google Colab will be used as the primary development environment, ensuring seamless collaboration and easy cloud-based execution. If deployed, the models can be containerized using Docker and deployed on cloud platforms for real-time predictions.

In conclusion, this Travel Capstone Project serves as a comprehensive exploration of how data can transform the travel industry. By integrating flight, hotel, and user data, the project demonstrates the power of data-driven decision-making. The insights gained will not only help businesses optimize their operations and improve customer experiences but also open opportunities for advanced applications like dynamic pricing, customer segmentation, and intelligent recommendation systems. This project bridges the gap between raw data and actionable intelligence, showcasing how data science can drive innovation in the travel and tourism sector.

**Merge Flights and Hotels**

````python
# Merge Flights and Hotels


# Step 1: Merge flights and hotels on 'travelCode' using inner join
flights_hotels = pd.merge(flights, hotels, on="travelCode", how="inner")

# Step 2: Merge flights_hotels with users
# Join flights_hotels.userCode_x with users.code (user ID)
travel_data = pd.merge(flights_hotels, users, left_on="userCode_x", right_on="code", how="inner")

# Step 3: Clean up columns
travel_data.rename(columns={'userCode_x': 'userCode'}, inplace=True)
if 'userCode_y' in travel_data.columns:
    travel_data.drop(columns=['userCode_y'], inplace=True)

# Final check
print("Shape of final merged dataframe:", travel_data.shape)
print(travel_data.info())
````

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Convert date columns to datetime
flights['date'] = pd.to_datetime(flights['date'])
hotels['date'] = pd.to_datetime(hotels['date'])

# Convert categorical columns to category dtype for efficiency
flights[['from', 'to', 'flightType', 'agency']] = flights[['from', 'to', 'flightType', 'agency']].apply(lambda x: x.astype('category'))
hotels[['name', 'place']] = hotels[['name', 'place']].apply(lambda x: x.astype('category'))
users[['company', 'gender']] = users[['company', 'gender']].apply(lambda x: x.astype('category'))

# Create new features: month and day of week from date
flights['flight_month'] = flights['date'].dt.month
flights['flight_dayofweek'] = flights['date'].dt.dayofweek

hotels['hotel_month'] = hotels['date'].dt.month
hotels['hotel_dayofweek'] = hotels['date'].dt.dayofweek

# Show first few rows of each dataset to verify changes
print("Flights dataset preview:")
print(flights.head())

print("\nHotels dataset preview:")
print(hotels.head())

print("\nUsers dataset preview:")
print(users.head())
````
