**Merge Flights and Hotels**

````python
# Merge Flights and Hotels


# Step 1: Merge flights and hotels on 'travelCode' using inner join
flights_hotels = pd.merge(flights, hotels, on="travelCode", how="inner")

# Step 2: Merge flights_hotels with users
# Join flights_hotels.userCode_x with users.code (user ID)
travel_data = pd.merge(flights_hotels, users, left_on="userCode_x", right_on="code", how="inner")

# Step 3: Clean up columns
travel_data.rename(columns={'userCode_x': 'userCode'}, inplace=True)
if 'userCode_y' in travel_data.columns:
    travel_data.drop(columns=['userCode_y'], inplace=True)

# Final check
print("Shape of final merged dataframe:", travel_data.shape)
print(travel_data.info())
````

## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Convert date columns to datetime
flights['date'] = pd.to_datetime(flights['date'])
hotels['date'] = pd.to_datetime(hotels['date'])

# Convert categorical columns to category dtype for efficiency
flights[['from', 'to', 'flightType', 'agency']] = flights[['from', 'to', 'flightType', 'agency']].apply(lambda x: x.astype('category'))
hotels[['name', 'place']] = hotels[['name', 'place']].apply(lambda x: x.astype('category'))
users[['company', 'gender']] = users[['company', 'gender']].apply(lambda x: x.astype('category'))

# Create new features: month and day of week from date
flights['flight_month'] = flights['date'].dt.month
flights['flight_dayofweek'] = flights['date'].dt.dayofweek

hotels['hotel_month'] = hotels['date'].dt.month
hotels['hotel_dayofweek'] = hotels['date'].dt.dayofweek

# Show first few rows of each dataset to verify changes
print("Flights dataset preview:")
print(flights.head())

print("\nHotels dataset preview:")
print(hotels.head())

print("\nUsers dataset preview:")
print(users.head())
````

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Answer Here.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.



**Hypothesis 1: Relationship Between Flight Distance and Price**


**Null Hypothesis** (H₀): There is no significant correlation between flight distance and flight price.

**Alternative Hypothesis** (H₁): There is a significant positive correlation between flight distance and flight price.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

from scipy.stats import pearsonr

corr_coef, p_value = pearsonr(flights['distance'], flights['price'])
print('Correlation Coefficient:', corr_coef)
print('P-Value:', p_value)
````

##### Which statistical test have you done to obtain P-Value?

Answer Here.

* The Pearson correlation test.


##### Why did you choose the specific statistical test?

Answer Here.

* Because both variables, distance and price, are continuous numerical variables.

* The Pearson correlation test quantifies the linear relationship between two continuous variables and provides a p-value to assess the significance of this relationship.

* It is the most appropriate and standard method to test whether a significant correlation exists between flight distance and price based on the data type and research question.

### Hypothetical Statement - 2

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Answer Here.

**Hypothesis 2: Difference in Average Flight Price Between Flight Types**

**Null Hypothesis (H₀):** The mean flight price is the same across all flight types (Economy, Business, First Class).

**Alternative Hypothesis (H₁):** At least one flight type has a significantly different mean price compared to the others.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value

from scipy.stats import kruskal

# Check sample sizes for each flight type
print(flights['flightType'].value_counts())

# Extract price data for each flight type group with correct category names
first_class_prices = flights[flights['flightType'] == 'firstClass']['price']
premium_prices = flights[flights['flightType'] == 'premium']['price']
economic_prices = flights[flights['flightType'] == 'economic']['price']

# Perform Kruskal-Wallis test
stat, p_value = kruskal(first_class_prices, premium_prices, economic_prices)

print('Kruskal-Wallis Statistic:', stat)
print('P-Value:', p_value)

# Interpretation
if p_value < 0.05:
    print("Reject Null Hypothesis: Significant difference in average flight price between flight types.")
else:
    print("Fail to Reject Null Hypothesis: No significant difference in average flight price between flight types.")

````

##### Which statistical test have you done to obtain P-Value?

Answer Here.

The statistical test used to obtain the p-value is the **Kruskal-Wallis test**. This test is a non-parametric method used to compare the medians of three or more independent groups, which in your case are different flight types (firstClass, premium, and economic). It is chosen because it does not assume a normal distribution of the data, making it suitable when the assumptions for ANOVA are not met or when sample sizes vary across groups. The p-value from this test indicates whether there is a statistically significant difference in flight prices among the flight types.


##### Why did you choose the specific statistical test?

Answer Here.

The Kruskal-Wallis test was chosen because it compares multiple groups without assuming normality, making it ideal for analyzing flight price differences across flight types.

### Hypothetical Statement - 3

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Answer Here.

**Hypothesis 3: Difference in Average Age Between Male and Female Users**

* **Null Hypothesis (H₀):** There is no significant difference in the average age between male and female users.

* **Alternative Hypothesis (H₁):** There is a significant difference in the average age between male and female users.

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value


from scipy.stats import mannwhitneyu

# Step 1: Clean and standardize gender values
users['gender'] = users['gender'].str.lower().str.strip()

# Step 2: Check group sizes
print(users['gender'].value_counts())

# Step 3: Separate age data for male and female users
male_age = users[users['gender'] == 'male']['age']
female_age = users[users['gender'] == 'female']['age']

# Step 4: Perform Mann–Whitney U Test
stat, p_value = mannwhitneyu(male_age, female_age, alternative='two-sided')

print('Mann–Whitney U Statistic:', stat)
print('P-Value:', p_value)

# Step 5: Interpretation
if p_value < 0.05:
    print("Reject Null Hypothesis: Significant difference in user age between genders.")
else:
    print("Fail to Reject Null Hypothesis: No significant difference in user age between genders.")
````

##### Which statistical test have you done to obtain P-Value?

Answer Here.

The p-value was obtained using the **Mann–Whitney U test**. It compares age between two groups (male and female) without assuming normal distribution. This test is suitable when data is non-normal or sample sizes differ.

##### Why did you choose the specific statistical test?

Answer Here.

The Mann-Whitney U test was chosen because it compares two independent groups (male and female) on a continuous variable (age) without assuming normal data distribution. It's ideal when data might be non-normal or sample sizes are unequal, which often occurs in real-world datasets. Unlike a parametric t-test, it ranks values and compares medians, making it robust against outliers. This ensures a reliable assessment of whether the average ages differ by gender, fitting your dataset's characteristics and hypothesis testing requirements effectively.

````python
# Check if any value is missing (True/False)

flights.isnull().values.any()  # Returns True if any missing value exists
hotels.isnull().values.any()
users.isnull().values.any()


flights.isnull().any()
hotels.isnull().any()
users.isnull().any()
````

````python
# Handling Missing Values & Missing Value Imputation

print(travel_data.isnull().sum())

def impute_missing_values(df, numeric_mean_cols=[], numeric_median_cols=[], categorical_mode_cols=[]):
    df = df.copy()

    for col in numeric_mean_cols:
        if df[col].isnull().sum() > 0:
            mean_val = df[col].mean()
            df[col].fillna(mean_val, inplace=True)
            print(f"Filled missing values in '{col}' with mean: {mean_val}")
        else:
            print(f"No missing values in '{col}'")

    for col in numeric_median_cols:
        if df[col].isnull().sum() > 0:
            median_val = df[col].median()
            df[col].fillna(median_val, inplace=True)
            print(f"Filled missing values in '{col}' with median: {median_val}")
        else:
            print(f"No missing values in '{col}'")

    for col in categorical_mode_cols:
        if df[col].isnull().sum() > 0:
            mode_val = df[col].mode()[0]
            df[col].fillna(mode_val, inplace=True)
            print(f"Filled missing values in '{col}' with mode: {mode_val}")
        else:
            print(f"No missing values in '{col}'")

    return df

travel_data = impute_missing_values(
    travel_data,
    numeric_mean_cols=['age', 'price_x', 'price_y'],
    numeric_median_cols=['distance'],
    categorical_mode_cols=['gender', 'flightType', 'agency']
)
````

#### What all missing value imputation techniques have you used and why did you use those techniques?

Answer Here.


1. Mean Imputation: For numerical columns such as 'age', 'price_x', 'price_y', where missing values are replaced with the column mean. This method is chosen due to its simplicity and effectiveness when data is approximately normally distributed and the missingness is low.

2. Median Imputation: For numerical columns like 'distance' where data may be skewed or contain outliers, median imputation offers a robust alternative that minimizes distortion from extreme values.

3. Mode Imputation: For categorical columns such as 'gender', 'flightType', and 'agency', missing values were filled with the most frequent category (mode), preserving the categorical distribution without introducing bias.

These techniques were selected because they are straightforward, computationally efficient, and suitable when missingness is relatively small or random. More advanced methods like K-Nearest Neighbors (KNN) or Multiple Imputation by Chained Equations (MICE) might be used if missing data is extensive or patterns are complex, but in this case, since missing values were minimal or absent, simple imputation sufficed.

By using these tailored imputation methods depending on data type and distribution, the project ensured data integrity and readiness for downstream analysis and model building without introducing significant bias or complexity. This balanced approach optimizes both performance and interpretability in data preprocessing.
